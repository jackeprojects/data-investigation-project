# What the data says

The table *movies* has `3201` rows and `16` columns, while the table *genre_groups* has `12` rows and `2` columns.

### Dtypes
Only one column needs to convert its dtype and it's inside the *movies* table:

- *Release Date* is a `str`, and should be converted to `DateTime`

Changing it to DateTime makes it easier to handle, like sorting by date instead of it being sorted alphabetically.

### What one row is
In *movies.csv*, one row is one movie with its budget, gross, release date, genre and ratings.

In *genre_groups*, one row is one genre and the group it belongs to.

### Missing data
The top 3 most empty columns are:
- *US DVD Sales*: 82.4% missing
- *Running Time min*: 62.2% missing
- *Director*: 41.6% missing

For example, 82.4% of movies are missing US DVD Sales. It seems unlikely that so many movies sold nothing at all. This is something I don't know, so I will have to investigate further.

## Cleaning the data

### Release Date
First I'll convert all dates into DateTime, then I will check if/how many conversions failed.

```python
# Convert movies to DateTime
movies["Release Date"] = pd.to_datetime(movies["Release Date"], format="%b %d %Y", errors="coerce")

# Check how many conversions failed
print(movies["Release Date"].isna().sum())
```

#### Result
All `3201` dates were converted successfully, none became `NaT`.
